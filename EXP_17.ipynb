```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Hp\Desktop\Salary_data_converted.csv")

print(df.head())
print(df.info())
print(df.describe())

df.dropna(inplace=True)

features = df.iloc[:, [0]].values
label = df.iloc[:, 1].values

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2, random_state=42
)

model = LinearRegression()
model.fit(x_train, y_train)

print("Training Score:", model.score(x_train, y_train))
print("Testing Score:", model.score(x_test, y_test))

y_pred = model.predict(x_test)

plt.scatter(features, label)
plt.plot(features, model.predict(features), color='red')
plt.title('Years of Experience vs Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()

print("Predicted Salary for 5 Years Experience:",
      model.predict([[5]])[0])
```
