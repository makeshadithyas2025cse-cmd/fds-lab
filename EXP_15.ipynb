```python
import numpy as np
import pandas as pd

# Read the newly saved Data.csv
df = pd.read_csv(r"C:\\Users\\Hp\\Desktop\\Data.csv")

print("Original Dataset:")
print(df)

# 1. Impute missing values
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

# 2. One-hot encode categorical feature
country_dummies = pd.get_dummies(df['Country'], dtype=int)

# 3. Concatenate feature columns
updated_dataset = pd.concat(
    [country_dummies, df[['Age', 'Salary', 'Purchased']]], 
    axis=1
)

# 4. Map target variable
updated_dataset['Purchased'] = updated_dataset['Purchased'].map({'No': 0, 'Yes': 1})

print("\nPreprocessed Dataset:")
print(updated_dataset)
```
