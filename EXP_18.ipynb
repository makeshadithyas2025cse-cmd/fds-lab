```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"C:\Users\Hp\Desktop\iris.csv")

print(df.head())
print(df.info())

features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

xtrain, xtest, ytrain, ytest = train_test_split(
    features, label, test_size=0.2, random_state=42
)

model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain, ytrain)

print("Training Score:", model_KNN.score(xtrain, ytrain))
print("Testing Score:", model_KNN.score(xtest, ytest))

from sklearn.metrics import confusion_matrix
print("Confusion Matrix:")
print(confusion_matrix(label, model_KNN.predict(features)))

from sklearn.metrics import classification_report
print("Classification Report:")
print(classification_report(label, model_KNN.predict(features)))
```
