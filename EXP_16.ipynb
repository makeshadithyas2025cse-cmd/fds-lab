```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read the newly saved CSV file
df = pd.read_csv(r"C:\Users\Hp\Desktop\EDA.csv")

# Clean column headers
df.columns = df.columns.str.strip().str.lower()

# Ensure numeric values
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['income'] = pd.to_numeric(df['income'], errors='coerce')

# 1. Summary Statistics
print("First 5 Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

# 2. Age Distribution
plt.figure(figsize=(6, 4))
plt.hist(df['age'].dropna(), bins=10, edgecolor='black', alpha=0.7)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.tight_layout()
plt.show()

# 3. Income Distribution
plt.figure(figsize=(6, 4))
plt.hist(df['income'].dropna(), bins=10, edgecolor='black', color='green', alpha=0.7)
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.title("Income Distribution")
plt.tight_layout()
plt.show()

# 4. Age vs Income Scatter Plot
plt.figure(figsize=(6, 4))
plt.scatter(df['age'], df['income'], color='purple', alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Income")
plt.title("Age vs Income")
plt.tight_layout()
plt.show()

# 5. Income Boxplot
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['income'], color='orange')
plt.title("Income Box Plot")
plt.tight_layout()
plt.show()

# 6. Correlation Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(df[['age', 'income']].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
